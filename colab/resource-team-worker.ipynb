# R0ice Resource Team -- Colab worker

Lend a Google Colab GPU to the [R0ice Resource Team](https://r0ice.com/resource-team/). It reads public-domain books, rewords principles from them, and runs small jobs summarizing public developer documentation and discussions. It only downloads public text, runs a model on Colab's GPU, and sends the results back.

**How to run:** type your name in the box below, then **Runtime → Run all** (Ctrl+F9). The T4 GPU is already preset.

It installs Ollama, pulls the model, and keeps starting new rounds until `hours` is used up, grabbing the latest worker script each round. Colab disconnects idle tabs, so click in the tab now and then; if it disconnects, just Run all again.

Worker source: https://r0ice.com/cloud-gpu-worker.py

```python
#@title Resource Team worker { display-mode: "form" }
your_name = "" #@param {type:"string"}
hours = 4 #@param {type:"number"}

import math, re, shlex, time
name = re.sub(r"[^A-Za-z0-9 _.-]", "", your_name).strip()[:40]
if not name:
    raise SystemExit("Type your name in the your_name box above, then Run all again.")
handle = f"Colab GPU ({name})"
ROUND_MIN = 235
rounds = max(1, math.ceil(hours * 60 / ROUND_MIN))
print(f"Joining as {handle} for about {hours} hours ({rounds} rounds)")
for i in range(rounds):
    print(f"--- round {i + 1}/{rounds} ({ROUND_MIN} min) ---", flush=True)
    get_ipython().system("curl -fsSL https://r0ice.com/cloud-gpu-worker.py -o worker.py")
    get_ipython().system("python3 worker.py --handle " + shlex.quote(handle) + " --minutes " + str(ROUND_MIN))
    time.sleep(5)
print("All rounds finished. Run all again to keep going.")
```
