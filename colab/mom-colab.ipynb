**Just Runtime → Run all (Ctrl+F9).** The T4 GPU and the worker name `Colab GPU (Mom)` are preset.

Free Colab gives a limited amount of GPU time and disconnects idle tabs, so click in the tab now and then. If it disconnects or says no GPU is available, try again later.

Volunteers: use [resource-team-worker.ipynb](https://colab.research.google.com/github/SethR0ice/r0ice-community-reader/blob/main/colab/resource-team-worker.ipynb) instead, so you run under your own name.

```python
#@title Resource Team worker { display-mode: "form" }
handle = "Colab GPU (Mom)" #@param {type:"string"}
hours = 4 #@param {type:"number"}

import math, shlex, time
ROUND_MIN = 235
rounds = max(1, math.ceil(hours * 60 / ROUND_MIN))
safe_handle = shlex.quote(handle.replace('"', '').strip())
# Private-lane token (2026-09-30): the server no longer hands this seat its token by name.
# Add a Colab secret named RT_TOKEN (key icon, left bar) and allow this notebook to read it.
# Without it the worker still runs, as a public-only identity.
token_arg = ""
try:
    from google.colab import userdata
    _t = (userdata.get("RT_TOKEN") or "").strip()
    if _t:
        token_arg = " --token=" + shlex.quote(_t)
except Exception:
    pass
for i in range(rounds):
    print(f"--- round {i + 1}/{rounds} ({ROUND_MIN} min) ---", flush=True)
    get_ipython().system("curl -fsSL https://r0ice.com/cloud-gpu-worker.py -o worker.py")
    get_ipython().system("python3 worker.py --handle " + safe_handle + " --minutes " + str(ROUND_MIN) + token_arg)
    time.sleep(5)
print("All rounds finished. Run all again to keep going.")
```
